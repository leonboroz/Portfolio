```python
import csv
import random
from datetime import datetime, timedelta
import calendar

# Константа для XOR-ключа — можно менять, чтобы менять кодировку
SECRET_KEY = 0x5A3C7F

names = [
    "Иван Иванов", "мария петрова", "Сергей Сидоров", "Анна Смирнова", "Ольга Кузнецова",
    "михаил орлов", "ЕЛЕНА НОВИКОВА", "Алексей Михайлов", "Наталья Васильева", "Дмитрий Кузнецов"
]

order_statuses = ['completed', 'processing', 'cancelled', 'Completed', 'Processing', 'Cancelled']

comments = [
    "спасибо за заказ", "Ждём с нетерпением!", "Задержка с доставкой, печалька...", 
    "Всё отлично!", "подарок вовремя", "Проверить наличие", "Нужно уточнить сроки"
]

products = [
    {"product_id": 1, "name": "Корм для собак", "category": "Товары для животных", "price": 500},
    {"product_id": 2, "name": "Игрушка мяч", "category": "Товары для животных", "price": 300},
    {"product_id": 3, "name": "Амуниция для собак", "category": "Товары для животных", "price": 1200},
    {"product_id": 4, "name": "Подписка на газету", "category": "Подписки", "price": 600},
    {"product_id": 5, "name": "Молоко", "category": "Продукты", "price": 3000},
]

def random_date_in_month(year, month):
    start = datetime(year, month, 1)
    last_day = calendar.monthrange(year, month)[1]
    end = datetime(year, month, last_day, 23, 59, 59)
    delta = end - start
    random_second = random.randint(0, int(delta.total_seconds()))
    date = start + timedelta(seconds=random_second)
    formats = ["%Y-%m-%d %H:%M:%S", "%d.%m.%Y %H:%M", "%Y/%m/%d %H:%M:%S"]
    fmt = random.choice(formats)
    return date.strftime(fmt)

def random_phone():
    formats = ["+7 9{}{}{} {}{} {}{}", "+7(9{}{}{}){}{}-{}{}", "8(9{}{}{}) {}{}-{}{}"]
    fmt = random.choice(formats)
    digits = [str(random.randint(0,9)) for _ in range(fmt.count('{}')*2)]
    return fmt.format(*digits)

def random_case(s):
    return ''.join(c.upper() if random.random() > 0.5 else c.lower() for c in s)

def encode_order_id(year, month, order_num):
    base_num = (year * 100 + month) * 10000 + order_num  # например 2024050001
    xored = base_num ^ SECRET_KEY
    random_suffix = random.randint(100, 999)  # случайный 3-значный суффикс
    return f"{xored}{random_suffix}"

def generate_order_positions(year, month, order_num):
    num_positions = random.randint(1, 5)  # позиций в заказе от 1 до 5
    positions = []
    order_date = random_date_in_month(year, month)
    delivery_date = random_date_in_month(year, month)[:10]
    status = random.choice(order_statuses)
    comment = random.choice(comments)
    customer_name = random_case(random.choice(names))
    customer_phone = random_phone()
    gift_card = ''.join(str(random.randint(0, 9)) for _ in range(16))
    order_id = encode_order_id(year, month, order_num)

    for _ in range(num_positions):
        product = random.choice(products)
        quantity = random.randint(1, 5)
        price = product['price']
        total_amount = round(price * quantity * (1 - random.uniform(0, 0.2)), 2)
        discount_amount = round(price * quantity - total_amount, 2)

        positions.append({
            "order_id": order_id,
            "order_date": order_date,
            "customer_name": customer_name,
            "customer_phone": customer_phone,
            "product_id": product["product_id"],
            "product_name": random_case(product["name"]),
            "quantity": quantity,
            "total_amount": total_amount,
            "discount_amount": discount_amount,
            "status": random_case(status),
            "comment": comment,
            "delivery_date": delivery_date,
            "gift_card": gift_card
        })
    return positions

header = [
    "order_id", "order_date", "customer_name", "customer_phone", "product_id", "product_name", 
    "quantity", "total_amount", "discount_amount", "status", "comment", "delivery_date", "gift_card"
]

def generate_csv_for_month(filename, year, month, num_orders=2000):
    with open(filename, mode='w', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=header)
        writer.writeheader()
        for order_num in range(1, num_orders + 1):
            positions = generate_order_positions(year, month, order_num)
            for pos in positions:
                writer.writerow(pos)

if __name__ == "__main__":
    for month in range(1, 13):
        filename = f"orders_2024_{month:02d}.csv"
        print(f"Генерируем файл {filename} с 2000 заказами...")
        generate_csv_for_month(filename, 2024, month, num_orders=2000)
    print("Генерация завершена.")
```
