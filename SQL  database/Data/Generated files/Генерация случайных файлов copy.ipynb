```python
import os
import csv
import random
from datetime import datetime, timedelta
import calendar
import string

# Папка для файлов
output_folder = "generated_files"
os.makedirs(output_folder, exist_ok=True)

# Пример простейшей генерации "левого" файла
def generate_random_file(filename, lines=100):
    with open(filename, 'w', encoding='utf-8') as f:
        for _ in range(lines):
            # Просто случайный текст с цифрами
            line = ''.join(random.choices(string.ascii_letters + string.digits + " ,.-", k=50))
            f.write(line + "\n")

# Твой основной генератор заказов (упрощённый, можно подставить свой)
def generate_csv_for_month(filename, year, month, num_orders=10):
    header = ["order_id", "order_date", "customer_name", "product_id", "quantity"]
    with open(filename, mode='w', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=header)
        writer.writeheader()
        for order_num in range(1, num_orders + 1):
            writer.writerow({
                "order_id": f"{year}{month:02d}{order_num:04d}",
                "order_date": f"{year}-{month:02d}-{random.randint(1,28)}",
                "customer_name": f"User{random.randint(1,100)}",
                "product_id": random.randint(1,5),
                "quantity": random.randint(1,10),
            })

# Создаем нужные файлы
for month in range(1, 4):  # например, 3 месяца
    filename = os.path.join(output_folder, f"orders_2024_{month:02d}.csv")
    generate_csv_for_month(filename, 2024, month, num_orders=1000)

# Создаем несколько "левых" файлов
for i in range(5):
    fname = os.path.join(output_folder, f"random_file_{i}_{random.randint(1000,9999)}.txt")
    generate_random_file(fname, lines=200)

print(f"Файлы созданы в папке {output_folder}")
```
